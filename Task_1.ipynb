```python
# Load the Dataset
df = pd.read_excel("Dataset for Data Analytics.xlsx")
# View the Dataset
print(df.shape)
df.head(10)
```

```python
# Check Column Information
df.info()
```

```python
# Find Missing (Null) Values
df.isnull().sum()
```

```python
# check what percentage is missing
missing = df["CouponCode"].isnull().sum()
total = len(df)

print("Missing values:", missing)
print("Percentage:", round(missing/total*100,2), "%")
```

```python
df["CouponCode"].unique()
```

```python
df["CouponCode"].head(20)
```

```python
# Handle missing values by filling with "No Coupon"
df["CouponCode"] = df["CouponCode"].fillna("No Coupon")
df["CouponCode"].isnull().sum()
```

The CouponCode column contained 309 missing values (25.75% of the dataset). Examination of the available values showed that the column stores promotional codes such as SAVE10, FREESHIP, and WINTER15. Since coupon usage is optional, missing values were interpreted as transactions where no coupon was applied. Therefore, missing values were replaced with "No Coupon" to preserve all transaction records while eliminating null values.Since coupon codes are optional promotional identifiers and a missing value likely indicates that no coupon was used during the transaction, the missing values were replaced with "No Coupon". This approach preserved all transaction records while eliminating null values.
